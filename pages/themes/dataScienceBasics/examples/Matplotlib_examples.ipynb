# Matplotlib Demos

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
csv_file = "https://raw.githubusercontent.com/ProgressBG-Python-Course/ProgressBG-Python-Slides/refs/heads/gh-pages/pages/themes/dataScienceBasics/datasets/developers.csv"

# Read the CSV data into a Pandas DataFrame
df = pd.read_csv(csv_file)
df
```

```python
# Set a nicer style for all plots
plt.style.use('seaborn-v0_8')
```

## Simple Bar Chart - Developer Salaries

```python
plt.figure(figsize=(8, 4))
plt.bar(df['Developer'], df['Salary'], color='skyblue')
plt.title('Simple Bar Chart - Developer Salaries', fontsize=16)
plt.xlabel('Developer', fontsize=12)
plt.ylabel('Salary ($)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
```

## Horizontal Bar Chart - Sorted by Salary

```python
plt.figure(figsize=(6, 4))
df_sorted = df.sort_values('Salary')
plt.barh(df_sorted['Developer'], df_sorted['Salary'], color='lightgreen')
plt.title('Developers Ranked by Salary', fontsize=16)
plt.xlabel('Salary ($)', fontsize=12)
plt.ylabel('Developer', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
```

## Scatter Plot - Age vs Salary

```python
plt.figure(figsize=(8, 4))
plt.scatter(df['Age'], df['Salary'], s=100, alpha=0.7, c=df['Age'], cmap='viridis')
plt.title('Age vs Salary Correlation', fontsize=16)
plt.xlabel('Age (years)', fontsize=12)
plt.ylabel('Salary ($)', fontsize=12)
plt.grid(linestyle='--', alpha=0.7)

# Add a trend line
z = np.polyfit(df['Age'], df['Salary'], 1)
p = np.poly1d(z)
plt.plot(df['Age'], p(df['Age']), "r--", alpha=0.8)
plt.show()
```

## Line Plot - Salary vs Age (sorted by age)

```python
plt.figure(figsize=(8, 4))
df_age_sorted = df.sort_values('Age')
plt.plot(df_age_sorted['Age'], df_age_sorted['Salary'], marker='o',
         linestyle='-', linewidth=2, markersize=8, color='purple')
plt.title('Salary Trend with Age', fontsize=16)
plt.xlabel('Age (years)', fontsize=12)
plt.ylabel('Salary ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
```

## Pie Chart - Age Distribution

```python
plt.figure(figsize=(8, 4))
age_bins = [20, 25, 30, 35, 45]
age_labels = ['20-24', '25-29', '30-34', '35+']

df['Age Group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)
age_counts = df['Age Group'].value_counts()

plt.pie(age_counts, labels=age_counts.index, autopct='%1.1f%%',
        startangle=90, shadow=True, explode=[0.05]*len(age_counts),
        colors=plt.cm.Pastel1.colors)
plt.title('Developer Age Distribution', fontsize=16)

plt.axis('equal')
plt.tight_layout()
```

## Save to file

```python
plt.figure()  # Explicitly create a figure
plt.hist(df['Age'], bins=5)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Developer Age Distribution")

plt.savefig("Developer Age Distribution.png")  # Save before closing
plt.close()
```
