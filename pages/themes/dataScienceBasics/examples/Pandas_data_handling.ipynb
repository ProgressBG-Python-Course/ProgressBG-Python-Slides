# Data Handling with Pandas


In next examples we'll be using a simple moc dataset [developers.csv](https://github.com/ProgressBG-Python-Course/ProgressBG-Python-Slides/blob/gh-pages/pages/themes/dataScienceBasics/datasets/developers.csv)

```python
import pandas as pd
```

## Load the data

Note that pandas has many functions for I/O from various file formats: [Pandas IO tools](https://pandas.pydata.org/docs/user_guide/io.html)
### From CSV

```python
csv_file = "https://raw.githubusercontent.com/ProgressBG-Python-Course/ProgressBG-Python-Slides/refs/heads/gh-pages/pages/themes/dataScienceBasics/datasets/developers.csv"

# Read the CSV data into a Pandas DataFrame
df = pd.read_csv(csv_file)
df
```

### From Excel (.xlsx)

```python
xlsx_file = "https://github.com/ProgressBG-Python-Course/ProgressBG-Python-Slides/raw/refs/heads/gh-pages/pages/themes/dataScienceBasics/datasets/developers.xlsx"

# Read the CSV data into a Pandas DataFrame
df = pd.read_excel(xlsx_file)
df
```

### From HTML

```python
tables = pd.read_html('https://pandas.pydata.org/docs/user_guide/io.html')
print(f"Found {len(tables)} tables!")
df_pandas_io = tables[0]
df_pandas_io
```

## Selecting Columns and Rows

```python
# Select a single column
df['Age']
```

```python
# Select multiple columns
df[['Age', 'Salary']]
```

```python
# Select rows by index
print(df.iloc[0]) # First row
print()
print(df.iloc[0:5]) # First 5 rows
```

## Sorting and Filtering Data

```python
df.sort_values(by='Salary', ascending=False)
```

```python
# Select rows based on condition
df[df['Age'] >= 30]
```

```python
# Filter data based on multiple conditions
mean_salary = df['Salary'].mean()
df[(df['Age'] <= 35) & (df['Salary'] >= mean_salary)]
```

## Save and export data 

### To CSV

```python
# 1. Basic CSV export
df.to_csv('developers_basic.csv', index=False)
print("Basic CSV export completed: developers_basic.csv")

# 2. CSV with custom separator and formatting
df.to_csv('developers_custom.csv',
          index=False,
          sep=';',                # Use semicolon as separator
          decimal=',',            # Use comma as decimal separator (European format)
          encoding='utf-8')       # Specify encoding
print("Custom CSV export completed: developers_custom.csv")
```

### To Excel (.xlsx)

Note, that the openpyxl module is needed when working with Excel (.xlsx) files in Pandas.
So, make sure that you have it: `pip install openpyxl`

```python
# 1. Basic Excel export
df.to_excel('developers_basic.xlsx', index=False)
print("Basic Excel export completed: developers_basic.xlsx")

# 2. Excel with sheet name specified
df.to_excel('developers_named_sheet.xlsx',
            sheet_name='Developer_Data',
            index=False)
print("Excel with named sheet export completed: developers_named_sheet.xlsx")
```
