# Data Science Basics Examples

```python
# needed modules

import numpy as np
import matplotlib.pyplot as plt
```

## Numpy
### Create a NumPy array

```python
array = np.array([1, 2, 3, 4, 5])
array
```

### Basic operations

```python
array + 5  # Element-wise addition
array * 2  # Element-wise multiplication
```

## Matplotlib

```python
# Simple line plot
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y)
plt.title('Square Numbers')
plt.xlabel('Number')
plt.ylabel('Square')
plt.show()
```
